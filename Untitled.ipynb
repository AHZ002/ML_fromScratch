```python
import numpy as np
import random
```

```python
np.random.seed(42)

N = 1000
X = np.random.rand(N , 1)
y = (np.sinh(X) > 0.5).astype(np.float32)
```

```python
init_scale = 0.01

W1 = np.random.randn(1, 4) * init_scale
b1 = np.zeros((1, 4))                  
W2 = np.random.randn(4, 1) * init_scale
b2 = np.zeros((1, 1)) 
```

```python
def sigmoid(z) : 

    return 1/(1 + np.exp(-z))

def sigmoid_derivative(x) :
    
    return x*(1-x)
```

```python
lr = 0.5             
max_epochs = 10000
tol = 1e-6           

prev_loss = float('inf')
for epoch in range(1, max_epochs + 1):
    
    z1 = X.dot(W1) + b1           
    a1 = sigmoid(z1)              
    z2 = a1.dot(W2) + b2          
    a2 = sigmoid(z2)              

    
    eps = 1e-8
    loss = -np.mean(y * np.log(a2 + eps) + (1 - y) * np.log(1 - a2 + eps))

    
    print(f"Epoch {epoch:5d} — loss: {loss:.6f}")

    
    if abs(prev_loss - loss) < tol:
        print(f"\nConverged at epoch {epoch}, loss = {loss:.6f}")
        break
    prev_loss = loss

    #backprop 
    
    dz2 = a2 - y                        
    dW2 = a1.T.dot(dz2) / N                
    db2 = np.sum(dz2, axis=0, keepdims=True) / N

    da1 = dz2.dot(W2.T)
    dz1 = da1 * sigmoid_derivative(a1)   
    dW1 = X.T.dot(dz1) / N               
    db1 = np.sum(dz1, axis=0, keepdims=True) / N

    W2 -= lr * dW2
    b2 -= lr * db2
    W1 -= lr * dW1
    b1 -= lr * db1

preds = (a2 >= 0.5).astype(int)
accuracy = np.mean(preds == y)
print(f"\nTraining accuracy: {accuracy * 100:.2f}%")
```
