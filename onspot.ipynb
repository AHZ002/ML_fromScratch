```python
#give python code for jupyter notebook to implement from scratch a neural network with 1 input neuron , 1 hidden layer with 4 neurons and output layer has 1 neuron , use synthetic data , do baclkpropagation , the network should converge after some converge ,   use the best weight initialization stratergy use cross entropy as loss function
```

```python
import numpy as np

# 1) Create very simple synthetic data (linearly separable via a threshold)
np.random.seed(42)
N = 1000
X = np.random.rand(N, 1)                    # inputs ∈ [0,1)
y = (X > 0.5).astype(np.float32)            # label = 1 if x>0.5 else 0

# 2) Random weight initialization (small Gaussian)
init_scale = 0.01
W1 = np.random.randn(1, 4) * init_scale     # weights for input → hidden
b1 = np.zeros((1, 4))                       # hidden biases
W2 = np.random.randn(4, 1) * init_scale     # weights for hidden → output
b2 = np.zeros((1, 1))                       # output bias

# 3) Activation & derivative
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

def sigmoid_derivative(a):
    return a * (1 - a)

# 4) Training settings
lr = 0.5             # learning rate
max_epochs = 10000
tol = 1e-6           # stop when loss change < tol

prev_loss = float('inf')
for epoch in range(1, max_epochs + 1):
    # --- Forward pass ---
    z1 = X.dot(W1) + b1           # (N,4)
    a1 = sigmoid(z1)              # (N,4)
    z2 = a1.dot(W2) + b2          # (N,1)
    a2 = sigmoid(z2)              # (N,1)  network output

    # --- Compute binary cross‐entropy loss ---
    eps = 1e-8
    loss = -np.mean(y * np.log(a2 + eps) + (1 - y) * np.log(1 - a2 + eps))

    # --- Print intermediate loss ---
    print(f"Epoch {epoch:5d} — loss: {loss:.6f}")

    # --- Convergence check ---
    if abs(prev_loss - loss) < tol:
        print(f"\nConverged at epoch {epoch}, loss = {loss:.6f}")
        break
    prev_loss = loss

    # --- Backpropagation ---
    dz2 = a2 - y                           # (N,1)
    dW2 = a1.T.dot(dz2) / N                # (4,1)
    db2 = np.sum(dz2, axis=0, keepdims=True) / N

    da1 = dz2.dot(W2.T)                    # (N,4)
    dz1 = da1 * sigmoid_derivative(a1)     # (N,4)
    dW1 = X.T.dot(dz1) / N                 # (1,4)
    db1 = np.sum(dz1, axis=0, keepdims=True) / N

    # --- Parameter update ---
    W2 -= lr * dW2
    b2 -= lr * db2
    W1 -= lr * dW1
    b1 -= lr * db1

else:
    print(f"\nDid not converge within {max_epochs} epochs, final loss = {loss:.6f}")

# 5) Evaluate on training set
preds = (a2 >= 0.5).astype(int)
accuracy = np.mean(preds == y)
print(f"\nTraining accuracy: {accuracy * 100:.2f}%")
```
