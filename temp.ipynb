```python
# Step 1: Define a function to calculate Euclidean distance between two points
def euclidean_distance(point1, point2):
    """
    Calculate the Euclidean distance between two points.
    Arguments:
    point1, point2: Lists containing the feature values of the points.
    
    Returns:
    The Euclidean distance.
    """
    distance = 0
    for i in range(len(point1)):
        distance += (point1[i] - point2[i]) ** 2
    return distance ** 0.5  # Return the square root of the sum

# Step 2: Define a function to compute weight (inverse of distance)
def calculate_weight(distance):
    """
    Calculate the weight inversely proportional to the distance.
    Arguments:
    distance: The Euclidean distance between the test point and a training point.
    
    Returns:
    The weight, where weight = 1 / distance.
    """
    if distance == 0:
        return float('inf')  # Handle division by zero (exact match case)
    return 1 / distance

# Step 3: Define a function to find the k nearest neighbors
def get_k_nearest_neighbors(X_train, y_train, test_point, k):
    """
    Get the k nearest neighbors for a given test point.
    
    Arguments:
    X_train: List of training points.
    y_train: List of corresponding labels.
    test_point: The test point whose neighbors need to be found.
    k: Number of nearest neighbors to consider.
    
    Returns:
    A list of the k nearest neighbors as (distance, label) tuples.
    """
    distances = []  # List to store (distance, label) tuples
    for i in range(len(X_train)):
        distance = euclidean_distance(X_train[i], test_point)
        distances.append((distance, y_train[i]))
    
    # Sort based on distance and select the k nearest neighbors
    distances.sort(key=lambda x: x[0])  # Sort by distance
    return distances[:k]

# Step 4: Define a function to predict the label based on the weighted KNN approach
def predict_label(X_train, y_train, test_point, k):
    """
    Predict the label for a given test point using weighted KNN.
    
    Arguments:
    X_train: List of training points.
    y_train: List of corresponding labels.
    test_point: The test point whose label needs to be predicted.
    k: Number of nearest neighbors to consider.
    
    Returns:
    The predicted label for the test point.
    """
    # Get the k nearest neighbors
    nearest_neighbors = get_k_nearest_neighbors(X_train, y_train, test_point, k)
    
    # Calculate the weighted vote for each class
    class_weights = {}  # Dictionary to store total weight for each class
    for distance, label in nearest_neighbors:
        weight = calculate_weight(distance)
        if label in class_weights:
            class_weights[label] += weight
        else:
            class_weights[label] = weight
    
    # Return the class with the highest total weight
    return max(class_weights, key=class_weights.get)

# Step 5: Define a function to predict labels for multiple test points
def predict_multiple(X_train, y_train, X_test, k):
    """
    Predict the labels for multiple test points using weighted KNN.
    
    Arguments:
    X_train: List of training points.
    y_train: List of corresponding labels.
    X_test: List of test points.
    k: Number of nearest neighbors to consider.
    
    Returns:
    A list of predicted labels for the test points.
    """
    predictions = []
    for test_point in X_test:
        predictions.append(predict_label(X_train, y_train, test_point, k))
    return predictions

# Step 6: Get input data from the user
# Input training data
X_train = []  # List to store feature values for training points
y_train = []  # List to store labels for training points

# Number of features in the dataset
num_features = int(input("Enter the number of features: "))

# Number of training points
n_train = int(input("Enter the number of training points: "))

# Input each training point's feature values and label
print("\nEnter the feature values and label for each training point (space-separated):")
for i in range(n_train):
    input_data = input(f"Training Point {i + 1}: ").split()
    features = list(map(float, input_data[:num_features]))  # Take first n elements as features
    label = int(input_data[num_features])  # Last element as label
    X_train.append(features)
    y_train.append(label)

# Input test data
X_test = []  # List to store feature values for test points
n_test = int(input("\nEnter the number of test points: "))

# Input each test point's feature values
print("\nEnter the feature values for each test point (space-separated):")
for i in range(n_test):
    features = list(map(float, input(f"Test Point {i + 1}: ").split()))  # Take all elements as features
    X_test.append(features)

# Step 7: Set the number of neighbors (k)
k = int(input("\nEnter the value of k (number of neighbors to consider): "))

# Step 8: Predict the labels for the test points
predicted_labels = predict_multiple(X_train, y_train, X_test, k)

# Step 9: Print the test data and the predicted labels
print("\nTest Data Points and their Predicted Labels:")
for i, test_point in enumerate(X_test):
    print(f"Test Point {i + 1} ({test_point}): Predicted Label = {predicted_labels[i]}")
```
