```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)
```

```python
# Initialize the network parameters
np.random.seed(42)
input_size = 3  # Example input features
hidden1_size = 7  # First hidden layer (7 neurons)
hidden2_size = 5  # Second hidden layer (5 neurons)
output_size = 2  # Two output units

# Random weight initialization
w1 = np.random.randn(input_size, hidden1_size)
b1 = np.random.randn(hidden1_size)
w2 = np.random.randn(hidden1_size, hidden2_size)
b2 = np.random.randn(hidden2_size)
w3 = np.random.randn(hidden2_size, output_size)
b3 = np.random.randn(output_size)
```

```python
x_train = np.array([[0.5, 0.3, 0.8]])  # Input
y_train = np.array([[1, 0]])  # Target output

learning_rate = 0.01
num_epochs = 2
```

```python
for epoch in range(num_epochs):
    for x, y in zip(x_train, y_train):
        # Forward propagation
        z1 = np.dot(x, w1) + b1
        a1 = sigmoid(z1)
        
        z2 = np.dot(a1, w2) + b2
        a2 = sigmoid(z2)
        
        z3 = np.dot(a2, w3) + b3
        a3 = sigmoid(z3)
        
        # Compute squared error
        error = 0.5 * np.sum((y - a3) ** 2)
        
        # Backpropagation
        d3 = (a3 - y) * sigmoid_derivative(a3)
        dw3 = np.dot(a2.reshape(-1, 1), d3.reshape(1, -1))
        db3 = d3
        
        d2 = np.dot(w3, d3) * sigmoid_derivative(a2)
        dw2 = np.dot(a1.reshape(-1, 1), d2.reshape(1, -1))
        db2 = d2
        
        d1 = np.dot(w2, d2) * sigmoid_derivative(a1)
        dw1 = np.dot(x.reshape(-1, 1), d1.reshape(1, -1))
        db1 = d1
        
        # Update weights and biases
        w3 -= learning_rate * dw3
        b3 -= learning_rate * db3
        
        w2 -= learning_rate * dw2
        b2 -= learning_rate * db2
        
        w1 -= learning_rate * dw1
        b1 -= learning_rate * db1
        
        print(f"Epoch {epoch+1}, Error: {error:.4f}")
```

```python
print("Final Weights and Biases:")
print("w1:", w1)
print("b1:", b1)
print("w2:", w2)
print("b2:", b2)
print("w3:", w3)
print("b3:", b3)
```
