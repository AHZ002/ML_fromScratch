```python
import numpy as np
import matplotlib.pyplot as plt


def perceptron_train(X, y, epochs=100):
    weights = np.zeros(X.shape[1])  
    bias = 0  
    
    for epoch in range(epochs):
        updates = 0
        for i in range(len(X)):
            
            activation = np.dot(X[i], weights) + bias
            
            prediction = 1 if activation >= 0 else 0
            
            if prediction != y[i]:
                update = y[i] - prediction
                weights += update * X[i]
                bias += update
                updates += 1
        
        if updates == 0:
            print(f"Converged after {epoch + 1} epochs")
            break
    return weights, bias

def perceptron_predict(X, weights, bias):
    predictions = []
    for x in X:
        activation = np.dot(x, weights) + bias
        predictions.append(1 if activation >= 0 else 0)
    return np.array(predictions)


X_and = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_and = np.array([0, 0, 0, 1])  


X_or = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_or = np.array([0, 1, 1, 1])  


weights_and, bias_and = perceptron_train(X_and, y_and)
weights_or, bias_or = perceptron_train(X_or, y_or)


plt.figure(figsize=(12, 6))


plt.subplot(1, 2, 1)
plt.scatter(X_and[:, 0], X_and[:, 1], c=y_and, cmap='bwr', edgecolor='k')
x_vals_and = np.linspace(-0.5, 1.5, 100)
y_vals_and = -(weights_and[0] * x_vals_and + bias_and) / weights_and[1]
plt.plot(x_vals_and, y_vals_and, color='black', label='Decision Boundary')
plt.title("AND Function Decision Boundary")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid()


plt.subplot(1, 2, 2)
plt.scatter(X_or[:, 0], X_or[:, 1], c=y_or, cmap='bwr', edgecolor='k')
x_vals_or = np.linspace(-0.5, 1.5, 100)
y_vals_or = -(weights_or[0] * x_vals_or + bias_or) / weights_or[1]
plt.plot(x_vals_or, y_vals_or, color='black', label='Decision Boundary')
plt.title("OR Function Decision Boundary")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]
labels = [0, 1, 1, 0]

def step_function(x):
    return 1 if x >= 0 else 0

# Neural Network Weights and Biases
w1 = [0.5, -0.5]  # Weights for first hidden neuron...
b1 = -1       # Bias for first hidden neuron..

w2 = [-0.5, 0.5]  # Weights for second hidden neuron....
b2 = -0.5         # Bias for second hidden neuron

w_out = [0.5, 0.5]  # Weights for output neuron
b_out = -1        # Bias for output neuron

# Forward pass for the neural network
def forward(x1, x2):
    # Hidden layer activations
    h1 = step_function(w1[0] * x1 + w1[1] * x2 + b1)
    h2 = step_function(w2[0] * x1 + w2[1] * x2 + b2)
    
    # Output layer activation
    y = step_function(w_out[0] * h1 + w_out[1] * h2 + b_out)
    return y

# Predict and compare with expected labels
print("Predictions for XOR function:")
for i in range(len(data)):
    x1, x2 = data[i]
    prediction = forward(x1, x2)
    print(f"Input: ({x1}, {x2}) → Predicted: {prediction} (Expected: {labels[i]})")

# Visualization of decision regions
x1_values = np.linspace(-0.5, 1.5, 100)
x2_values = np.linspace(-0.5, 1.5, 100)
grid = np.array([[x1, x2] for x1 in x1_values for x2 in x2_values])

# Predictions for the grid points
predictions = np.array([forward(x1, x2) for x1, x2 in grid])
predictions = predictions.reshape(len(x1_values), len(x2_values))

# Plotting the decision regions
plt.contourf(
    x1_values,
    x2_values,
    predictions,
    levels=[-0.5, 0.5, 1.5],
    colors=['blue', 'red'],
    alpha=0.3
)


for i, (x1, x2) in enumerate(data):
    color = 'red' if labels[i] == 1 else 'blue'
    plt.scatter(x1, x2, color=color, s=100, edgecolors='black')


plt.xlabel('x1')
plt.ylabel('x2')
plt.title('XOR Function Decision Regions')
plt.xlim(-0.5, 1.5)
plt.ylim(-0.5, 1.5)
plt.grid(True)


plt.show()
```
